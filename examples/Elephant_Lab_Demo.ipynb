# Elephant Lab Demo
### An interactive guide to exploring electrophysiology data with Elephant Lab

This notebook walks you through the core features of **Elephant Lab** — a JupyterLab extension for interactive exploration and visualization of electrophysiology data.

We will use a real monkey reach-to-grasp dataset (`i140703-001_no_raw.nix`) to demonstrate each feature step by step.

---
## Step 1 — Open Elephant Lab
> <span style="color:red">IMPORTANT:</span> Make sure you open a Notebook first as Elephant Lab depends on an active Notebook!.


Open the **Command Palette** on either via `View -> Active Command Palette` or press `Ctrl+Shift+C`.

Search for **`Elephant Lab`** and click it.

The **Elephant Lab panel** will open and show:
- **Top half**: Neo object tree
- **Bottom half**: Neo object **Details** Panel and **Explore** Panel (rasterplot for SpikeTrains, LFP plot for AnalogSignals)

---
## Step 2 — Load a Dataset via the Load Button

> <span style="color:orange">Warning:</span> Loading a dataset might take a while especially for large datasets!

In the Elephant Lab panel, click the **Load** button.

A file browser will appear. Navigate to and select:

```
examples/i140703-001_no_raw.nix
```

The Neo object tree will populate with the dataset hierarchy. You should see a `Block` node at the top level of the tree.

---
## Step 3 — The Block and the Details Panel

Click on the **`Block`** node at the top of the tree.

Because `Block` is a **container**, clicking it automatically selects it together with **all its descendants** — every Segment, AnalogSignal, SpikeTrain, Event, and Epoch inside it.

In the **Details** tab you will therefore see a **grouped Overview** of all selected objects, broken down by type (e.g. *SpikeTrain Overview (N)*, *AnalogSignal Overview (N)*, …).

The `Block` is the top-level container in the Neo data model — it groups all segments (trials) and their signals from a single recording session.

> **Tip:** To inspect the metadata of a single object (name, annotations, time range, …) click directly on a **leaf node** such as a SpikeTrain or AnalogSignal instead or deselect the children by Ctrl+Clicking them.

---
## Step 4 — The Segment and the Details Panel

Expand the `Block` node by clicking the arrow next to it, then click on one of the **`Segment`** nodes.

Like `Block`, a `Segment` is a **container**. Clicking it automatically selects it together with **all its children** — every AnalogSignal, SpikeTrain, Event, and Epoch belonging to that trial.

In the **Details** tab you will therefore see a **grouped Overview** of all the signal objects in that segment, broken down by type.

Each `Segment` represents one trial or recording epoch. Expanding it reveals all the signal objects it contains.

> **Tip:** Click on an individual child node (e.g. a SpikeTrain or AnalogSignal) to see the metadata for that single object.

---
## Step 5 — AnalogSignal: Details Panel

Expand a `Segment` node and click on one of the **`AnalogSignal`** entries.

In the **`Details`** tab you will see:
- Signal name and units (e.g. `mV`)
- Sampling rate
- Duration and time range
- Number of channels
- Any annotations attached to the signal (electrode position, channel ID, etc.)

AnalogSignals in this dataset are **LFP (Local Field Potential)** recordings.

---
## Step 6 — AnalogSignal: Explore

With the same `AnalogSignal` still selected, switch to the **`Explore`** tab in the Node Explorer.

You will see:
- A **time-series plot** of the raw signal
- Interactive zoom and pan controls
- Options to customize the plot to your needs

The overview LFP plot is also visible in the **bottom half** of the Elephant Lab panel.

---
## Step 7 — SpikeTrain: Details- and Explore Panel 

Within the same `Segment`, click on one of the **`SpikeTrain`** nodes.

**Details Panel** (`Details` tab):
- Unit name and ID
- `t_start` / `t_stop` of the recording
- Number of spikes
- Annotations such as `SNR`, `sua` (single-unit activity flag), `electrode_id`, and quality metrics

**Explore** (`Explore` tab):
- Rasterplot of selected SpikeTrain

The **rasterplot** overview in the bottom panel also highlights the selected unit.

---
## Step 8 — Event: Details- and Explore Panel 

Within the same `Segment`, click on one of the **`Event`** nodes.

**Details Panel** (`Details` tab):
- Event name
- Number of events and their labels (e.g. `TS-ON`, `GO-ON`, `SR`, `STOP`)
- Time stamps
- Annotations describing trial conditions

**Explore** (`Explore` tab):
- An event timeline showing all event markers

Events in this dataset mark **trial phases** of the reach-to-grasp task (target onset, go cue, movement start, etc.).

---
## Step 9 — Multi-Selection: Event and SpikeTrain

Hold `Ctrl` (or `Cmd` on macOS) and click both an **`Event`** node **and** a **`SpikeTrain`** node to select them together.

**Details Panel**: shows a combined view of both objects.

**Explore**: displays a joint view — the spike raster aligned with the event markers on the same time axis, making it easy to see how spikes relate to task events.

---
## Step 10 — Multiple SpikeTrains: Explore and Annotations

Hold `Ctrl` (or `Cmd` on macOS) and click **several `SpikeTrain` nodes** to build a multi-unit selection.

**Details** (`Details` tab):
- Overview of the SpikeTrains
- Overview of the Annotations

**Explore Panel** (`Explore` tab):
- Plot the SpikeTrains next to each other in order to compare them
> **Tip:** You might want to click on _Overlap_ to compare them

---
## Step 11 — Insert Code

With one or more objects still selected in the tree, click the **`Insert`** button in the Elephant Lab toolbar.

Elephant Lab will automatically generate a Python variable at the current mouse position in the active notebook cell, referencing the selected Neo objects by name. This lets you seamlessly transition from visual exploration to programmatic analysis without having to manually look up variable names or indices.

> **Tip:** The generated code respects the selection — a single SpikeTrain generates a single-object snippet, while a multi-unit selection generates a list.

---
## Step 12 — Create an AnalogSignal from Scratch

You can also create Neo objects directly in Python and Elephant Lab will pick them up automatically and add them to the tree.

Run the cell below to create a synthetic `AnalogSignal`, then watch it appear as a new node in the Elephant Lab tree view.

````python
import numpy as np
import quantities as pq
from neo import AnalogSignal

# Create a simple synthetic LFP-like signal (1 second, 1 kHz, 2 channels)
duration = 1.0  # seconds
sampling_rate = 1000 * pq.Hz
t = np.linspace(0, duration, int(duration * sampling_rate.magnitude))

channel_1 = np.sin(2 * np.pi * 10 * t)   # 10 Hz oscillation
channel_2 = np.sin(2 * np.pi * 40 * t)   # 40 Hz oscillation

signal_data = np.column_stack([channel_1, channel_2]) * pq.mV

my_signal = AnalogSignal(
    signal_data,
    sampling_rate=sampling_rate,
    name="My Synthetic LFP",
    description="Two-channel synthetic signal: 10 Hz and 40 Hz oscillations"
)

print(my_signal)
````
