# Elephant Lab Demo
### An interactive guide to exploring electrophysiology data with Elephant Lab

This notebook walks you through the core features of **Elephant Lab** — a JupyterLab extension for interactive exploration and visualization of electrophysiology data.

We will use a real monkey reach-to-grasp dataset (`i140703-001_no_raw.nix`) to demonstrate each feature step by step.

---
## Step 1 — Open Elephant Lab
> <span style="color:red">IMPORTANT:</span> Make sure a Notebook is already open — Elephant Lab requires an active notebook kernel!

Open the **Command Palette** via `View → Activate Command Palette` or press `Ctrl+Shift+C`.

Search for **`Elephant Lab`** and click it.

The **Elephant Lab panel** will open on the right and show:
- **Top**: Neo object tree with a filter toolbar
- **Bottom**: **Details** tab and **Explore** tab (switchable)

---
## Step 1b — Download the Demo Dataset

Before loading data you need the file `i140703-001_no_raw.nix` (~740 MB).

Run the code cell below — it automatically picks the right command for your OS:

| Platform | Tool used |
|---|---|
| **Linux** | `wget` |
| **macOS** | `curl` (pre-installed) |
| **Windows** | PowerShell `Invoke-WebRequest` |

> **Alternative — manual download:** Open the URL below in your browser and save the file into the same folder as this notebook:
> ```
> https://gin.g-node.org/INT/multielectrode_grasp/src/master/datasets_nix/i140703-001_no_raw.nix
> ```

````python
from urllib.request import urlretrieve
from tqdm.notebook import tqdm
from pathlib import Path

url = "https://gin.g-node.org/INT/multielectrode_grasp/raw/master/datasets_nix/i140703-001_no_raw.nix"
out = Path("./i140703-001_no_raw.nix")

if out.exists():
    print("File already exists, skipping download.")
else:
    with tqdm(unit="B", unit_scale=True, unit_divisor=1024, desc="Downloading") as t:
        def reporthook(block, block_size, total):
            if total > 0:
                t.total = total
            t.update(block * block_size - t.n)

        urlretrieve(url, out, reporthook=reporthook)
````

---
## Step 2 — Load a Dataset

> <span style="color:orange">Warning:</span> Loading can take a while for large datasets (~12 s for this file).

In the Elephant Lab toolbar, click the <i class="fa fa-file-import"></i> **Load** button.

A file browser will appear. Navigate to and select the downloaded file:

```
i140703-001_no_raw.nix
```

A dialog will then ask for the **Neo IO class** — this is the reader Neo uses to parse the file format.
Click **Automatic** and Elephant Lab will detect the right one for you (`NixIO` for `.nix` files).

> **What is a Neo IO?**  
> [Neo](https://neo-python.readthedocs.io/) uses IO classes to read different file formats:  
> `NixIO` → `.nix`, `BlackrockIO` → `.nev`/`.ns*`, `NeuralynxIO` → `.ncs`, and [many more](https://neo.readthedocs.io/en/latest/rawio.html).  
> Each IO knows the format's structure and returns the data as a standard Neo object hierarchy.

The tree will populate with the dataset. You should see a <i class="fa fa-cube"></i> `Block` node at the top level.

---
## Step 3 — The <i class="fa fa-cube"></i> Block and the Details Panel

**Single-click** the **<i class="fa fa-cube"></i> `Block`** node at the top of the tree.

In the **Details** tab you will see the Block's own metadata:
- Name and description
- Recording date and file origin
- Block-level annotations (session metadata, experimental conditions)

The <i class="fa fa-cube"></i> `Block` is the top-level container in the Neo data model — it groups all segments (trials) and their signals from a single recording session.

> **Tip — select the whole <i class="fa fa-cube"></i> Block at once:** **Double-click** the <i class="fa fa-cube"></i> `Block` node to select it together with *all* its children (every <i class="fa fa-columns"></i> Segment, <i class="fa fa-water"></i> AnalogSignal, <i class="fa fa-braille"></i> SpikeTrain, <i class="fa fa-map-marker"></i> Event, …). The Details tab will then show a **grouped Overview** broken down by object type.

---
## Step 4 — The <i class="fa fa-columns"></i> Segment and the Details Panel

Expand the <i class="fa fa-cube"></i> `Block` node by clicking the expand icon next to it.

**Single-click** one of the **<i class="fa fa-columns"></i> `Segment`** nodes.

In the **Details** tab you will see the Segment's metadata:
- Index and name
- Annotations describing the trial (condition, recording context, …)

Each <i class="fa fa-columns"></i> `Segment` represents one trial or recording epoch. Expanding it reveals all the signal objects it contains.

> **Tip — select the whole <i class="fa fa-columns"></i> Segment at once:** **Double-click** the <i class="fa fa-columns"></i> `Segment` node to select it together with all its children (<i class="fa fa-water"></i> AnalogSignals, <i class="fa fa-braille"></i> SpikeTrains, <i class="fa fa-map-marker"></i> Events, …). The Details tab will then show an **Overview** of all contained objects.

---
## Step 5 — AnalogSignal: Details Panel

Expand a <i class="fa fa-columns"></i> `Segment` node and **single-click** one of the **<i class="fa fa-water"></i> `AnalogSignal`** entries.

In the **Details** tab you will see:
- Signal name and units (e.g. `mV`)
- Sampling rate and time range
- Number of channels
- Annotations (electrode position, channel ID, etc.)

In this dataset each <i class="fa fa-water"></i> `AnalogSignal` entry represents a **multi-channel LFP (Local Field Potential) recording**.

---
## Step 6 — AnalogSignal: Explore

With the same <i class="fa fa-water"></i> `AnalogSignal` still selected, switch to the **`Explore`** tab.

You will see:
- A **time-series plot** of the LFP signal
- Interactive zoom and pan controls
- A toolbar with plot options:
  - <i class="fa fa-moon"></i> **Dark** — toggle dark / light theme
  - <i class="fa fa-layer-group"></i> **Overlap** — stack or overlay multiple signals
  - <i class="fa fa-compress"></i> **Normalize Y** — normalize signal amplitude
  - <i class="fa fa-expand-arrows-alt"></i> **Upscale** — re-render at higher resolution for the current zoom window

> **Note on downsampling:** For performance, long signals are automatically **downsampled** to a maximum of 10 000 points before plotting. Use <i class="fa fa-expand-arrows-alt"></i>**Upscale** to increase detail within your current zoom range.

---
## Step 7 — SpikeTrain: Details and Explore

Within the same <i class="fa fa-columns"></i> `Segment`, **single-click** one of the **<i class="fa fa-braille"></i> `SpikeTrain`** nodes.

**Details** tab:
- Unit name and ID
- `t_start` / `t_stop` of the recording
- Number of spikes
- Annotations: `SNR`, `sua` (single-unit activity flag), electrode quality metrics, …

**Explore** tab:
- Rasterplot of the selected SpikeTrain

---
## Step 8 — Event: Details and Explore

Within the same <i class="fa fa-columns"></i> `Segment`, **single-click** one of the **<i class="fa fa-map-marker"></i> `Event`** nodes.

**Details** tab:
- Event name
- Number of events and their labels (e.g. `TS-ON`, `GO-ON`, `SR`, `STOP`)
- Timestamps
- Annotations describing trial conditions

**Explore** tab:
- An event timeline showing all event markers

<i class="fa fa-map-marker"></i> Events in this dataset mark **trial phases** of the reach-to-grasp task (target onset, go cue, movement start, etc.).

---
## Step 9 — Multi-Selection: <i class="fa fa-map-marker"></i> Event and <i class="fa fa-braille"></i> SpikeTrain

You can add objects to the current selection without clearing it.

**Hold `Ctrl`** (or `Cmd` on macOS) and **single-click** a **<i class="fa fa-braille"></i> `SpikeTrain`** node to add it to the selection — then **Ctrl+click** an **<i class="fa fa-map-marker"></i> `Event`** node as well.

**Details** tab: shows a combined from all selected objects.

**Explore** tab: displays a joint view — the spike raster aligned with the event markers on the same time axis, making it easy to see how spikes relate to task events.

> **Note on Ctrl+click vs. Ctrl+double-click:**
> - `Ctrl+click` toggles **only the clicked node** in/out of the selection.
> - `Ctrl+double-click` toggles the clicked node **and all its children** — useful for adding an entire Segment's worth of signals to the selection at once.

---
## Step 10 — Multiple SpikeTrains: Explore and Annotations

There are two convenient ways to build a multi-unit selection:

**Option A — Ctrl+click individual units:**  
Hold `Ctrl` and single-click several **<i class="fa fa-braille"></i> `SpikeTrain`** nodes one by one.

**Option B — Shift+click for a range:**  
Click the first <i class="fa fa-braille"></i> SpikeTrain, then hold `Shift` and click another to select everything in between.  
Use `Shift+double-click` as the second click to also include all children of each node in the range.

**Details** tab:
- Per-unit annotations listed for each selected unit (SNR, `sua`, electrode quality flags)
- Lets you quickly compare unit quality across the selection before running analysis

**Explore** tab:
- Multi-unit rasterplot showing all selected spike trains on a shared time axis

> **Tip:** Toggle **Overlap** in the Explore toolbar to overlay all rasters on the same axis.

---
## Step 10b — The Filter Bar

The toolbar at the top of the Neo tree has two kinds of filtering:

### 1 — Object-type visibility
A row of icon toggles lets you **hide or show** each Neo object type in the tree. Click an icon to toggle it:

| Icon | Object type |
|---|---|
| <i class="fa fa-cube"></i> | Block |
| <i class="fa fa-columns"></i> | Segment |
| <i class="fa fa-water"></i> | AnalogSignal |
| <i class="fa fa-braille"></i> | SpikeTrain |
| <i class="fa fa-map-marker"></i> | Event |
| <i class="fa fa-hourglass"></i> | Epoch |
| <i class="fa fa-check"></i> | Expand / collapse all containers |

### 2 — Annotation expression filter
Below the type icons is a text input. Type a Python-style expression and press `Enter` (or click <i class="fa fa-filter"></i>):

```
sua==True AND spike_count>500
```

Elephant Lab evaluates the expression against every object's `.annotations` and **highlights all matches** in the tree, updating the Details and Explore views immediately.

Supported operators: `==`, `!=`, `>`, `<`, `>=`, `<=`, `AND`, `OR`.  
The input turns red if no objects match.

---
## Step 11 — Insert Code

With one or more objects selected in the tree, click the <i class="fa fa-code"></i> **Insert** button in the Elephant Lab toolbar.

Elephant Lab inserts a Python variable assignment **at the cursor position** in the currently active notebook cell, referencing the selected Neo objects by name. This lets you seamlessly transition from visual exploration to programmatic analysis.

> **Tip:** The generated code adapts to the selection — a single <i class="fa fa-braille"></i> SpikeTrain produces one variable, while a multi-unit selection produces a list.

## Step 12 — Create an <i class="fa fa-water"></i> AnalogSignal from Scratch

You can also create Neo objects directly in Python and Elephant Lab will pick them up automatically and add them to the tree.

Run the cell below to create a synthetic <i class="fa fa-water"></i> `AnalogSignal`, then watch it appear as a new node in the Elephant Lab tree view.

````python
import numpy as np
import quantities as pq
from neo import AnalogSignal

# Create a simple synthetic LFP-like signal (1 second, 1 kHz, 2 channels)
duration = 1.0  # seconds
sampling_rate = 1000 * pq.Hz
t = np.linspace(0, duration, int(duration * sampling_rate.magnitude))

channel_1 = np.sin(2 * np.pi * 10 * t)   # 10 Hz oscillation
channel_2 = np.sin(2 * np.pi * 40 * t)   # 40 Hz oscillation

signal_data = np.column_stack([channel_1, channel_2]) * pq.mV

my_signal = AnalogSignal(
    signal_data,
    sampling_rate=sampling_rate,
    name="My Synthetic LFP",
    description="Two-channel synthetic signal: 10 Hz and 40 Hz oscillations"
)

print(my_signal)
````
