# Elephant Lab Demo
### An interactive guide to exploring electrophysiology data with Elephant Lab

This notebook walks you through the core features of **Elephant Lab** — a JupyterLab extension for interactive exploration and visualization of electrophysiology data.

We will use a real monkey reach-to-grasp dataset (`i140703-001_no_raw.nix`) to demonstrate each feature step by step.

---
## Step 1 — Open Elephant Lab
> <span style="color:red">IMPORTANT:</span> Make sure you open a Notebook first as Elephant Lab depends on an active Notebook!


Open the **Command Palette** via `View -> Active Command Palette` or press `Ctrl+Shift+C`.

Search for **`Elephant Lab`** and click it.

The **Elephant Lab panel** will open and show:
- **Top**: Neo object tree
- **Bottom**: **Details** panel and **Explore** panel (as two tabs)

---
## Step 1b — Download the Demo Dataset

Before loading data you need the demo file `i140703-001_no_raw.nix`.
Download it with the command for your operating system, or run the code cell below directly from this notebook.

**Linux / macOS** (terminal or notebook cell with `!`):
```bash
wget -O examples/i140703-001_no_raw.nix "https://gin.g-node.org/INT/multielectrode_grasp/raw/master/datasets_nix/i140703-001_no_raw.nix"
```

> **Tip:** You can also just open the URL in your browser and save the file manually into the `examples/` folder next to this notebook.

---
## Step 2 — Load a Dataset via the Load Button

> <span style="color:orange">Warning:</span> Loading a dataset might take a while, especially for large datasets!

In the Elephant Lab panel, click the **Load** button.

A file browser will appear. Navigate to and select:

```
examples/i140703-001_no_raw.nix
```

A dialog will ask you to specify the **Neo IO class**:
- Click **Automatic** to let Elephant Lab detect the correct reader for you.
- Or type a class name manually (e.g. `NixIO`) and click **OK**.

The Neo object tree will populate with the dataset hierarchy. You should see a `Block` node at the top level of the tree.

---
## Step 3 — The Block and the Details Panel

**Single-click** the **`Block`** node at the top of the tree.

In the **Details** tab you will see the Block's own metadata:
- Name and description
- Recording date and file origin
- Block-level annotations (session metadata, experimental conditions)

The `Block` is the top-level container in the Neo data model — it groups all segments (trials) and their signals from a single recording session.

> **Tip — select the whole Block at once:** **Double-click** the `Block` node to select it together with *all* its children (every Segment, AnalogSignal, SpikeTrain, Event, …). The Details tab will then show a **grouped Overview** broken down by object type.

---
## Step 4 — The Segment and the Details Panel

Expand the `Block` node by clicking the **+** arrow next to it.

**Single-click** one of the **`Segment`** nodes.

In the **Details** tab you will see the Segment's metadata:
- Index and name
- Recording time range (`t_start`, `t_stop`)
- Annotations describing what happened during this segment

Each `Segment` represents one trial or recording epoch. Expanding it reveals all the signal objects it contains.

> **Tip — select the whole Segment at once:** **Double-click** the `Segment` node to select it together with all its children (AnalogSignals, SpikeTrains, Events, …). The Details tab will then show an **Overview** of all contained objects.

---
## Step 5 — AnalogSignal: Details Panel

Expand a `Segment` node and **single-click** one of the **`AnalogSignal`** entries.

In the **Details** tab you will see:
- Signal name and units (e.g. `mV`)
- Sampling rate
- Duration and time range
- Number of channels
- Annotations attached to the signal (electrode position, channel ID, etc.)

AnalogSignals in this dataset are **LFP (Local Field Potential)** recordings.

---
## Step 6 — AnalogSignal: Explore

With the same `AnalogSignal` still selected, switch to the **`Explore`** tab.

You will see:
- A **time-series plot** of the raw LFP signal
- Interactive zoom and pan controls
- Plot options (dark mode, overlap, normalization, …) in the toolbar above the plot

---
## Step 7 — SpikeTrain: Details and Explore

Within the same `Segment`, **single-click** one of the **`SpikeTrain`** nodes.

**Details** tab:
- Unit name and ID
- `t_start` / `t_stop` of the recording
- Number of spikes
- Annotations: `SNR`, `sua` (single-unit activity flag), electrode quality metrics, …

**Explore** tab:
- Rasterplot of the selected SpikeTrain

---
## Step 8 — Event: Details and Explore

Within the same `Segment`, **single-click** one of the **`Event`** nodes.

**Details** tab:
- Event name
- Number of events and their labels (e.g. `TS-ON`, `GO-ON`, `SR`, `STOP`)
- Timestamps
- Annotations describing trial conditions

**Explore** tab:
- An event timeline showing all event markers

Events in this dataset mark **trial phases** of the reach-to-grasp task (target onset, go cue, movement start, etc.).

---
## Step 9 — Multi-Selection: Event and SpikeTrain

You can add objects to the current selection without clearing it.

**Hold `Ctrl`** (or `Cmd` on macOS) and **single-click** a **`SpikeTrain`** node to add it to the selection — then **Ctrl+click** an **`Event`** node as well.

**Details** tab: shows a combined from all selected objects.

**Explore** tab: displays a joint view — the spike raster aligned with the event markers on the same time axis, making it easy to see how spikes relate to task events.

> **Note on Ctrl+click vs. Ctrl+double-click:**
> - `Ctrl+click` toggles **only the clicked node** in/out of the selection.
> - `Ctrl+double-click` toggles the clicked node **and all its children** — useful for adding an entire Segment's worth of signals to the selection at once.

---
## Step 10 — Multiple SpikeTrains: Explore and Annotations

There are two convenient ways to build a multi-unit selection:

**Option A — Ctrl+click individual units:**  
Hold `Ctrl` and single-click several **`SpikeTrain`** nodes one by one.

**Option B — Shift+click for a range:**  
Click the first SpikeTrain, then hold `Shift` and click another to select everything in between.  
Use `Shift+double-click` as the second click to also include all children of each node in the range.

**Details** tab:
- Per-unit annotations listed for each selected unit (SNR, `sua`, electrode quality flags)
- Lets you quickly compare unit quality across the selection before running analysis

**Explore** tab:
- Multi-unit rasterplot showing all selected spike trains on a shared time axis

> **Tip:** Toggle **Overlap** in the Explore toolbar to overlay all rasters on the same axis.

---
## Step 10b — Filter by Annotations

Instead of clicking nodes manually, use the **annotation filter** bar at the top of the Neo tree to select objects by their properties.

Type an expression into the filter input and press `Enter` (or click the filter button):

```
sua==True AND spike_count>500
```

Elephant Lab will highlight all matching SpikeTrains in the tree and immediately update the **Details** and **Explore** views for the filtered set.

Supported operators: `==`, `!=`, `>`, `<`, `>=`, `<=`, `AND`, `OR`.

---
## Step 11 — Insert Code

With one or more objects selected in the tree, click the **`Insert`** button in the Elephant Lab toolbar.

Elephant Lab inserts a Python variable assignment **at the cursor position** in the currently active notebook cell, referencing the selected Neo objects by name. This lets you seamlessly transition from visual exploration to programmatic analysis without having to manually look up variable names or indices.

> **Tip:** The generated code adapts to the selection — a single SpikeTrain produces a single variable, while a multi-unit selection produces a list.

---
## Step 12 — Create an AnalogSignal from Scratch

You can also create Neo objects directly in Python and Elephant Lab will pick them up automatically and add them to the tree.

Run the cell below to create a synthetic `AnalogSignal`, then watch it appear as a new node in the Elephant Lab tree view.

````python
import numpy as np
import quantities as pq
from neo import AnalogSignal

# Create a simple synthetic LFP-like signal (1 second, 1 kHz, 2 channels)
duration = 1.0  # seconds
sampling_rate = 1000 * pq.Hz
t = np.linspace(0, duration, int(duration * sampling_rate.magnitude))

channel_1 = np.sin(2 * np.pi * 10 * t)   # 10 Hz oscillation
channel_2 = np.sin(2 * np.pi * 40 * t)   # 40 Hz oscillation

signal_data = np.column_stack([channel_1, channel_2]) * pq.mV

my_signal = AnalogSignal(
    signal_data,
    sampling_rate=sampling_rate,
    name="My Synthetic LFP",
    description="Two-channel synthetic signal: 10 Hz and 40 Hz oscillations"
)

print(my_signal)
````
