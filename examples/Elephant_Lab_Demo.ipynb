# Elephant Lab Demo

## What is Elephant Lab?

Electrophysiological data from neuroscience experiments, such as spike train data or raw recorded voltage traces, are often highly complex. The unique nature of the experiments carried out in different projects, paired with the specific organization of data recorded from various recording systems and behavioral devices, using proprietary and custom file formats, leads to data bundles that are often difficult to load, understand and process.

The *Neo* Python library provides an interface that enables users to represent data from a large number of supported vendor-specific file formats in defined machine-actionable standard that is largely self-descriptive from its organizational structure elements and provided metadata.

**Elephant Lab** is a *JupyterLab* extension for the interactive exploration and visualization of electrophysiology data represented in *Neo*. It can be thought of as a specialized variable explorer that accompanies your live Jupyter coding and lets you inspect and visualize any *Neo* data objects that are loaded or generated. In this way, you will easily stay on top of your coding session and instantly explore the structure of novel datasets.

Future iterations to **Elephant Lab** will introduce analysis capabilities, so stay tuned.

**Using a command-line download tool**

Various tools for downloading files exist on the different platforms:
| Platform | Tool used |
|---|---|
| **Linux** | `wget` |

````python
from urllib.request import urlretrieve
from tqdm.notebook import tqdm
from pathlib import Path

url = "https://gin.g-node.org/INT/multielectrode_grasp/raw/master/datasets_nix/i140703-001_no_raw.nix"
out = Path("./i140703-001_no_raw.nix")

if out.exists():
    print("File already exists, skipping download.")
else:
    with tqdm(unit="B", unit_scale=True, unit_divisor=1024, desc="Downloading") as t:
        def reporthook(block, block_size, total):
            if total > 0:
                t.total = total
            t.update(block * block_size - t.n)

        urlretrieve(url, out, reporthook=reporthook)
````

### Annotation expression filters
Below the icons is a text input field. Type a Python-style expression and press `Enter` (or click <i class="fa fa-filter"></i>):

```
sua==True AND spike_count>500
```

**Elephant Lab** evaluates the expression against every **selected object's** standard properties (e.g., `name`, `description`, ...) and against its `annotations`. Positive matches are **highlighted** in the tree, updating the **`Details`** and **`Explore`** tabs accordingly.

Supported operators: `==`, `!=`, `>`, `<`, `>=`, `<=`, `AND`, `OR`.  

> <span style="color:orange">**Tip:** </span>The text input field is highlighted in red if no objects match the search criteria.

## Insight 11: Continuous Monitoring of New Neo Objects

You can also create *Neo* objects directly in Python and **Elephant Lab** will pick them up automatically and add them to the tree.

Run the cell below to create a synthetic **<i class="fa fa-water"></i> `AnalogSignal`**, then watch it appear as a new node in the Elephant Lab tree view.

````python
import numpy as np
import quantities as pq
from neo import AnalogSignal

# Create a simple synthetic LFP-like signal (1 second, 1 kHz, 2 channels)
duration = 1.0  # seconds
sampling_rate = 1000 * pq.Hz
t = np.linspace(0, duration, int(duration * sampling_rate.magnitude))

channel_1 = np.sin(2 * np.pi * 10 * t)   # 10 Hz oscillation
channel_2 = np.sin(2 * np.pi * 40 * t)   # 40 Hz oscillation

signal_data = np.column_stack([channel_1, channel_2]) * pq.mV

my_signal = AnalogSignal(
    signal_data,
    sampling_rate=sampling_rate,
    name="My Synthetic LFP",
    description="Two-channel synthetic signal: 10 Hz and 40 Hz oscillations"
)

print(my_signal)
````
