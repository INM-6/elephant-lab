# Jupyphant Neo Tree Test Notebook

This notebook is designed to test the reactivity of the Jupyphant's Neo object tree. Execute the cells one by one and observe the changes in the Jupyphant panel.

```python
import neo
import quantities as pq
import numpy as np
```

## 1. Create a Block

Execute the next cell to create a `neo.Block`. After execution, you should see a `Test Block` appear in the Jupyphant's Neo Tree view.

```python
# Create a new Block
my_block = neo.Block(name="Test Block")
```

## 2. Add a Segment

Now, let's add a `neo.Segment` to the block. Execute the following cell. You should see `Test Segment` appear under `Test Block` in the tree.

```python
# Create a new Segment and add it to the block
my_segment = neo.Segment(name="")
my_block.segments.append(my_segment)
```

## 3. Add Data to the Segment

Let's add some data to the segment. This will add a `SpikeTrain` and an `AnalogSignal` to the segment. You should see them appear in the tree under `Test Segment`.

```python
# Add some data to the segment to make it more interesting
st = neo.SpikeTrain(np.arange(0, 10) * pq.s, t_stop=10 * pq.s)
my_segment.spiketrains.append(st)

anasig = neo.AnalogSignal(np.random.randn(100, 1) * pq.mV, sampling_rate=10 * pq.Hz)
my_segment.analogsignals.append(anasig)
```

## 4. Remove the Segment

Now, execute the cell below to remove the segment from the block. The `Test Segment` should disappear from the tree.

```python
# Remove the segment
my_block.segments.remove(my_segment)
```

## 5. Add and Clear Segments

This cell will add a new segment. The next cell will clear all segments from the block.

```python
# You can also clear all segments
my_segment2 = neo.Segment(name="Another Segment")
my_block.segments.append(my_segment2)
# The next execution will clear it
```

After executing the cell above, `Another Segment` should be visible in the tree. Now, execute the cell below to clear it.

```python
my_block.segments.clear()
```

## 6. Deleting the Block

Finally, you can delete the block variable. This should remove it from the Neo tree.

```python
del my_block
```
